```python
import cloudpickle, pyspark.serializers
pyspark.serializers.cloudpickle = cloudpickle
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Ejercicio3_MapFilter").getOrCreate()
sc = spark.sparkContext

# Datos de entrada: 1 a 50
numeros = sc.parallelize(range(1, 51))

# 1. Multiplicar cada número por 3
multiplicados = numeros.map(lambda x: x * 3)

# 2. Filtrar solo los números mayores a 50
mayores_50 = multiplicados.filter(lambda x: x > 50)

# 3 y 4. Contar y mostrar resultados
total_mayores_50 = mayores_50.count()
primeros_10 = mayores_50.take(10)

print(f"Primeros 5 tras multiplicar por 3 : {multiplicados.take(5)} ... Último: {multiplicados.max()}")
print(f"Total de números mayores a 50     : {total_mayores_50}")
print(f"Primeros 10                       : {primeros_10}")

spark.stop()
```
