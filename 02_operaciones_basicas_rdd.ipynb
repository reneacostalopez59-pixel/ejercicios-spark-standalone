```python
import cloudpickle, pyspark.serializers
pyspark.serializers.cloudpickle = cloudpickle
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Ejercicio2_BasicasRDD").getOrCreate()
sc = spark.sparkContext

# Datos de entrada
numeros = list(range(1, 21))  # [1, 2, 3, ..., 20]

# 1. Crear RDD
rdd_numeros = sc.parallelize(numeros)

# 2, 3, 4 y 5. Operaciones básicas
total = rdd_numeros.count()
primeros_5 = rdd_numeros.take(5)
suma = rdd_numeros.sum()
maximo = rdd_numeros.max()
minimo = rdd_numeros.min()

print(f"Total de elementos : {total}")
print(f"Primeros 5         : {primeros_5}")
print(f"Suma               : {suma}")
print(f"Máximo: {maximo}, Mínimo: {minimo}")

spark.stop()
```
