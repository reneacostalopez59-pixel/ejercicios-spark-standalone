```python
import cloudpickle, pyspark.serializers
pyspark.serializers.cloudpickle = cloudpickle
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Ejercicio4_ClaveValor").getOrCreate()
sc = spark.sparkContext

# Datos de entrada
ventas = [
    ("Laptop", 1200),
    ("Mouse", 25),
    ("Laptop", 1100),
    ("Teclado", 75),
    ("Mouse", 30),
    ("Laptop", 1300)
]

# 1. Crear RDD
rdd_ventas = sc.parallelize(ventas)

# 2. Total de ventas por producto usando reduceByKey
total_por_producto = rdd_ventas.reduceByKey(lambda a, b: a + b)

# 3. Contar cuántas ventas hay de cada producto
conteo_por_producto = rdd_ventas.map(lambda x: (x[0], 1)).reduceByKey(lambda a, b: a + b)

# 4. Producto con mayor total de ventas
producto_top = total_por_producto.sortBy(lambda x: x[1], ascending=False).first()

print("Total por producto :", dict(total_por_producto.sortByKey().collect()))
print("Número de ventas   :", dict(conteo_por_producto.sortByKey().collect()))
print(f"Producto con mayor total: {producto_top[0]} ({producto_top[1]})")

spark.stop()
```
