```python
from pyspark.sql import SparkSession

# 1, 2 y 3. Crear SparkSession con configuración personalizada
spark = SparkSession.builder \
    .appName("MiPrimerAnalisis") \
    .config("spark.executor.memory", "1g") \
    .config("spark.sql.shuffle.partitions", "8") \
    .getOrCreate()

# 4. Verificar que la sesión se creó correctamente mostrando su información
print("Nombre de la app       :", spark.sparkContext.appName)
print("Versión de Spark       :", spark.version)
print("Master                 :", spark.sparkContext.master)
print("Memoria del executor   :", spark.conf.get("spark.executor.memory"))
print("Particiones de shuffle :", spark.conf.get("spark.sql.shuffle.partitions"))
print("\nObjeto SparkSession:")
display(spark)

# Cerrar sesión
spark.stop()
print("Sesión cerrada correctamente.")
```
